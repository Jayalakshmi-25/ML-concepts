```python
import pandas as pd
import numpy as np

# ============================================================
# 1. READ DATASET
# ============================================================

file_path = "/content/drive/MyDrive/Home Credit/application_train.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully")
print("Rows   :", df.shape[0])
print("Columns:", df.shape[1])


# ============================================================
# 2. FUNCTION TO CALCULATE MEMORY
# ============================================================

def get_memory_usage(df):
    """
    Returns total DataFrame memory usage in MB.
    """
    memory_bytes = df.memory_usage(deep=True).sum()
    memory_mb = memory_bytes / (1024 ** 2)

    return memory_mb


# ============================================================
# 3. MEMORY OPTIMIZATION FUNCTION
# ============================================================

def optimize_memory(df):
    """
    Reduce DataFrame memory usage by:
    1. Downcasting integer columns
    2. Downcasting float columns
    3. Converting suitable object columns to category
    """

    # Memory before optimization
    start_memory = get_memory_usage(df)

    print("\n" + "=" * 60)
    print("MEMORY OPTIMIZATION")
    print("=" * 60)

    print(f"Memory Before Optimization : {start_memory:.2f} MB")

    # Loop through every column
    for col in df.columns:

        col_type = df[col].dtype

        # ----------------------------------------------------
        # Optimize Integer Columns
        # ----------------------------------------------------

        if pd.api.types.is_integer_dtype(col_type):

            c_min = df[col].min()
            c_max = df[col].max()

            if c_min >= np.iinfo(np.int8).min and \
               c_max <= np.iinfo(np.int8).max:

                df[col] = df[col].astype(np.int8)

            elif c_min >= np.iinfo(np.int16).min and \
                 c_max <= np.iinfo(np.int16).max:

                df[col] = df[col].astype(np.int16)

            elif c_min >= np.iinfo(np.int32).min and \
                 c_max <= np.iinfo(np.int32).max:

                df[col] = df[col].astype(np.int32)

            else:

                df[col] = df[col].astype(np.int64)


        # ----------------------------------------------------
        # Optimize Float Columns
        # ----------------------------------------------------

        elif pd.api.types.is_float_dtype(col_type):

            c_min = df[col].min()
            c_max = df[col].max()

            if c_min >= np.finfo(np.float16).min and \
               c_max <= np.finfo(np.float16).max:

                df[col] = df[col].astype(np.float16)

            elif c_min >= np.finfo(np.float32).min and \
                 c_max <= np.finfo(np.float32).max:

                df[col] = df[col].astype(np.float32)

            else:

                df[col] = df[col].astype(np.float64)


        # ----------------------------------------------------
        # Optimize Object / String Columns
        # ----------------------------------------------------

        elif col_type == "object":

            unique_values = df[col].nunique(dropna=False)
            total_values = len(df[col])

            # Convert to category if less than 50%
            # of the values are unique
            if total_values > 0:

                unique_ratio = unique_values / total_values

                if unique_ratio < 0.50:
                    df[col] = df[col].astype("category")


    # --------------------------------------------------------
    # Memory After Optimization
    # --------------------------------------------------------

    end_memory = get_memory_usage(df)

    memory_saved = start_memory - end_memory

    reduction_percentage = (
        (memory_saved / start_memory) * 100
        if start_memory > 0
        else 0
    )

    print(f"Memory After Optimization  : {end_memory:.2f} MB")
    print(f"Memory Saved               : {memory_saved:.2f} MB")
    print(f"Memory Reduction           : {reduction_percentage:.2f}%")

    print("=" * 60)

    return df


# ============================================================
# 4. CHECK MEMORY BEFORE OPTIMIZATION
# ============================================================

memory_before = get_memory_usage(df)

print("\nBefore Optimization")
print(f"Memory Usage: {memory_before:.2f} MB")


# ============================================================
# 5. OPTIMIZE DATASET
# ============================================================

df = optimize_memory(df)


# ============================================================
# 6. CHECK MEMORY AFTER OPTIMIZATION
# ============================================================

memory_after = get_memory_usage(df)

print("\nAfter Optimization")
print(f"Memory Usage: {memory_after:.2f} MB")


# ============================================================
# 7. FINAL MEMORY SAVING
# ============================================================

memory_saved = memory_before - memory_after

percentage_saved = (
    (memory_saved / memory_before) * 100
    if memory_before > 0
    else 0
)

print("\n" + "=" * 60)
print("FINAL MEMORY REPORT")
print("=" * 60)

print(f"Before Optimization : {memory_before:.2f} MB")
print(f"After Optimization  : {memory_after:.2f} MB")
print(f"Memory Saved        : {memory_saved:.2f} MB")
print(f"Percentage Saved    : {percentage_saved:.2f}%")

print("=" * 60)


# ============================================================
# 8. DISPLAY OPTIMIZED DATA TYPES
# ============================================================

print("\nOptimized Data Types:")
print(df.dtypes.value_counts())
```

```python
import pandas as pd
import numpy as np

# ============================================================
# 1. READ DATASET
# ============================================================

file_path = "/content/drive/MyDrive/Home Credit/application_train.csv"

df = pd.read_csv(file_path)

print("Dataset Shape Before Cleaning:", df.shape)


# ============================================================
# 2. FUNCTION TO HANDLE MISSING VALUES
# ============================================================

def handle_missing_values(df, threshold=30):

    # Create copy
    df = df.copy()

    print("\n" + "=" * 70)
    print("MISSING VALUE HANDLING")
    print("=" * 70)

    # --------------------------------------------------------
    # STEP 1: Calculate Missing Percentage
    # --------------------------------------------------------

    missing_percentage = (df.isnull().sum() / len(df)) * 100

    # Only columns having missing values
    missing_percentage = missing_percentage[
        missing_percentage > 0
    ].sort_values(ascending=False)

    print("\nColumns having missing values:")
    print(missing_percentage)


    # --------------------------------------------------------
    # STEP 2: Find Columns >= 30% Missing
    # --------------------------------------------------------

    columns_to_drop = missing_percentage[
        missing_percentage >= threshold
    ].index.tolist()

    print("\n" + "=" * 70)
    print(f"COLUMNS WITH >= {threshold}% NULL VALUES")
    print("=" * 70)

    print("Number of columns to drop:", len(columns_to_drop))

    for col in columns_to_drop:
        print(
            col,
            "->",
            f"{missing_percentage[col]:.2f}% missing"
        )


    # --------------------------------------------------------
    # STEP 3: Drop Columns
    # --------------------------------------------------------

    df.drop(
        columns=columns_to_drop,
        inplace=True
    )

    print("\nDropped Columns:")
    print(columns_to_drop)


    # --------------------------------------------------------
    # STEP 4: Find Numerical and Categorical Columns
    # --------------------------------------------------------

    numerical_columns = df.select_dtypes(
        include=np.number
    ).columns

    categorical_columns = df.select_dtypes(
        include=["object", "category"]
    ).columns


    # --------------------------------------------------------
    # STEP 5: Fill Numerical Columns with Median
    # --------------------------------------------------------

    numerical_filled = []

    for col in numerical_columns:

        if df[col].isnull().sum() > 0:

            median_value = df[col].median()

            df[col] = df[col].fillna(median_value)

            numerical_filled.append(col)

            print(
                f"Numerical: {col} -> "
                f"filled with median = {median_value}"
            )


    # --------------------------------------------------------
    # STEP 6: Fill Categorical Columns with Mode
    # --------------------------------------------------------

    categorical_filled = []

    for col in categorical_columns:

        if df[col].isnull().sum() > 0:

            mode_values = df[col].mode(dropna=True)

            if not mode_values.empty:

                mode_value = mode_values.iloc[0]

                # Add category if necessary
                if isinstance(df[col].dtype, pd.CategoricalDtype):

                    if mode_value not in df[col].cat.categories:
                        df[col] = df[col].cat.add_categories(
                            [mode_value]
                        )

                df[col] = df[col].fillna(mode_value)

                categorical_filled.append(col)

                print(
                    f"Categorical: {col} -> "
                    f"filled with mode = {mode_value}"
                )


    # --------------------------------------------------------
    # STEP 7: Final Report
    # --------------------------------------------------------

    print("\n" + "=" * 70)
    print("FINAL CLEANING REPORT")
    print("=" * 70)

    print("Dropped Columns:")
    print(columns_to_drop)

    print("\nNumerical Columns Filled with Median:")
    print(numerical_filled)

    print("\nCategorical Columns Filled with Mode:")
    print(categorical_filled)

    print("\nDataset Shape After Cleaning:")
    print(df.shape)

    print("\nTotal Remaining Null Values:")
    print(df.isnull().sum().sum())

    print("=" * 70)

    return df, columns_to_drop, numerical_filled, categorical_filled


# ============================================================
# 3. CALL FUNCTION
# ============================================================

df_cleaned, dropped_columns, median_columns, mode_columns = \
    handle_missing_values(df, threshold=30)


# ============================================================
# 4. PRINT BEFORE AND AFTER SHAPE
# ============================================================

print("\nOriginal Dataset Shape :", df.shape)
print("Cleaned Dataset Shape  :", df_cleaned.shape)


# ============================================================
# 5. CHECK REMAINING NULL VALUES
# ============================================================

remaining_nulls = df_cleaned.isnull().sum()

remaining_nulls = remaining_nulls[
    remaining_nulls > 0
]

print("\nRemaining Columns with Null Values:")

if len(remaining_nulls) == 0:
    print("No missing values remaining.")
else:
    print(remaining_nulls)
```

```python
# =====================================================================
# HOME CREDIT - COMPLETE FEATURE ENGINEERING + CLASSIFICATION PIPELINE
# =====================================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, BaggingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)


# =====================================================================
# 1. START WITH CLEANED + ENCODED DATAFRAME
# =====================================================================

df_featured = df.copy()

eps = 1e-6

print("=" * 80)
print("FEATURE ENGINEERING")
print("=" * 80)

print("Dataset shape before feature engineering:", df_featured.shape)


# =====================================================================
# 2. CREATE 20 NEW FEATURES
# =====================================================================

# ---------------------------------------------------------------------
# FEATURE 1: Credit amount relative to applicant income
# ---------------------------------------------------------------------

df_featured["CREDIT_INCOME_RATIO"] = (
    df_featured["AMT_CREDIT"] /
    (df_featured["AMT_INCOME_TOTAL"] + eps)
)


# ---------------------------------------------------------------------
# FEATURE 2: Annuity/payment burden relative to income
# ---------------------------------------------------------------------

df_featured["ANNUITY_INCOME_RATIO"] = (
    df_featured["AMT_ANNUITY"] /
    (df_featured["AMT_INCOME_TOTAL"] + eps)
)


# ---------------------------------------------------------------------
# FEATURE 3: Credit amount relative to annuity
# Rough indicator of repayment duration
# ---------------------------------------------------------------------

df_featured["CREDIT_ANNUITY_RATIO"] = (
    df_featured["AMT_CREDIT"] /
    (df_featured["AMT_ANNUITY"] + eps)
)


# ---------------------------------------------------------------------
# FEATURE 4: Goods price relative to credit
# ---------------------------------------------------------------------

df_featured["GOODS_CREDIT_RATIO"] = (
    df_featured["AMT_GOODS_PRICE"] /
    (df_featured["AMT_CREDIT"] + eps)
)


# ---------------------------------------------------------------------
# FEATURE 5: Difference between credit and goods price
# ---------------------------------------------------------------------

df_featured["CREDIT_GOODS_DIFF"] = (
    df_featured["AMT_CREDIT"] -
    df_featured["AMT_GOODS_PRICE"]
)


# ---------------------------------------------------------------------
# FEATURE 6: Income available per family member
# ---------------------------------------------------------------------

df_featured["INCOME_PER_PERSON"] = (
    df_featured["AMT_INCOME_TOTAL"] /
    (df_featured["CNT_FAM_MEMBERS"] + eps)
)


# ---------------------------------------------------------------------
# FEATURE 7: Credit burden per family member
# ---------------------------------------------------------------------

df_featured["CREDIT_PER_PERSON"] = (
    df_featured["AMT_CREDIT"] /
    (df_featured["CNT_FAM_MEMBERS"] + eps)
)


# ---------------------------------------------------------------------
# FEATURE 8: Annuity per family member
# ---------------------------------------------------------------------

df_featured["ANNUITY_PER_PERSON"] = (
    df_featured["AMT_ANNUITY"] /
    (df_featured["CNT_FAM_MEMBERS"] + eps)
)


# ---------------------------------------------------------------------
# FEATURE 9: Percentage/proportion of family members who are children
# ---------------------------------------------------------------------

df_featured["CHILDREN_RATIO"] = (
    df_featured["CNT_CHILDREN"] /
    (df_featured["CNT_FAM_MEMBERS"] + eps)
)


# ---------------------------------------------------------------------
# FEATURE 10: Convert age from days into years
# ---------------------------------------------------------------------

df_featured["AGE_YEARS"] = (
    -df_featured["DAYS_BIRTH"] / 365.25
)


# ---------------------------------------------------------------------
# FEATURE 11: Employment duration in years
#
# Home Credit contains 365243 as a special/sentinel value.
# We convert it to NaN before calculating employment years.
# ---------------------------------------------------------------------

employment_days = df_featured["DAYS_EMPLOYED"].copy()

employment_days = employment_days.where(
    employment_days.abs() < 365243,
    np.nan
)

df_featured["EMPLOYMENT_YEARS"] = (
    -employment_days / 365.25
)


# ---------------------------------------------------------------------
# FEATURE 12: Employment duration relative to applicant age
# ---------------------------------------------------------------------

df_featured["EMPLOYMENT_AGE_RATIO"] = (
    df_featured["EMPLOYMENT_YEARS"] /
    (df_featured["AGE_YEARS"] + eps)
)


# ---------------------------------------------------------------------
# FEATURE 13: Registration duration converted to years
# ---------------------------------------------------------------------

df_featured["REGISTRATION_YEARS"] = (
    -df_featured["DAYS_REGISTRATION"] / 365.25
)


# ---------------------------------------------------------------------
# FEATURE 14: ID publication duration converted to years
# ---------------------------------------------------------------------

df_featured["ID_PUBLISH_YEARS"] = (
    -df_featured["DAYS_ID_PUBLISH"] / 365.25
)


# ---------------------------------------------------------------------
# FEATURE 15: Mean of external source scores
# ---------------------------------------------------------------------

df_featured["EXT_SOURCE_MEAN"] = (
    df_featured[
        [
            "EXT_SOURCE_1",
            "EXT_SOURCE_2",
            "EXT_SOURCE_3"
        ]
    ].mean(axis=1)
)


# ---------------------------------------------------------------------
# FEATURE 16: Minimum external source score
# ---------------------------------------------------------------------

df_featured["EXT_SOURCE_MIN"] = (
    df_featured[
        [
            "EXT_SOURCE_1",
            "EXT_SOURCE_2",
            "EXT_SOURCE_3"
        ]
    ].min(axis=1)
)


# ---------------------------------------------------------------------
# FEATURE 17: Maximum external source score
# ---------------------------------------------------------------------

df_featured["EXT_SOURCE_MAX"] = (
    df_featured[
        [
            "EXT_SOURCE_1",
            "EXT_SOURCE_2",
            "EXT_SOURCE_3"
        ]
    ].max(axis=1)
)


# ---------------------------------------------------------------------
# FEATURE 18: Standard deviation of external source scores
# ---------------------------------------------------------------------

df_featured["EXT_SOURCE_STD"] = (
    df_featured[
        [
            "EXT_SOURCE_1",
            "EXT_SOURCE_2",
            "EXT_SOURCE_3"
        ]
    ].std(axis=1)
)


# ---------------------------------------------------------------------
# FEATURE 19: 30-day social circle default ratio
# ---------------------------------------------------------------------

df_featured["SOCIAL_DEFAULT_30_RATIO"] = (
    df_featured["DEF_30_CNT_SOCIAL_CIRCLE"] /
    (
        df_featured["OBS_30_CNT_SOCIAL_CIRCLE"]
        + eps
    )
)


# ---------------------------------------------------------------------
# FEATURE 20: 60-day social circle default ratio
# ---------------------------------------------------------------------

df_featured["SOCIAL_DEFAULT_60_RATIO"] = (
    df_featured["DEF_60_CNT_SOCIAL_CIRCLE"] /
    (
        df_featured["OBS_60_CNT_SOCIAL_CIRCLE"]
        + eps
    )
)


# =====================================================================
# 3. LIST OF NEW FEATURES
# =====================================================================

new_features = [
    "CREDIT_INCOME_RATIO",
    "ANNUITY_INCOME_RATIO",
    "CREDIT_ANNUITY_RATIO",
    "GOODS_CREDIT_RATIO",
    "CREDIT_GOODS_DIFF",
    "INCOME_PER_PERSON",
    "CREDIT_PER_PERSON",
    "ANNUITY_PER_PERSON",
    "CHILDREN_RATIO",
    "AGE_YEARS",
    "EMPLOYMENT_YEARS",
    "EMPLOYMENT_AGE_RATIO",
    "REGISTRATION_YEARS",
    "ID_PUBLISH_YEARS",
    "EXT_SOURCE_MEAN",
    "EXT_SOURCE_MIN",
    "EXT_SOURCE_MAX",
    "EXT_SOURCE_STD",
    "SOCIAL_DEFAULT_30_RATIO",
    "SOCIAL_DEFAULT_60_RATIO"
]


print("\n20 New Features Created:")

for number, feature in enumerate(new_features, start=1):
    print(number, feature)


# =====================================================================
# 4. HANDLE INFINITY AND MISSING VALUES CREATED BY FEATURE ENGINEERING
# =====================================================================

df_featured.replace(
    [np.inf, -np.inf],
    np.nan,
    inplace=True
)


# Fill remaining numeric missing values with median

for col in df_featured.select_dtypes(include=np.number).columns:

    if df_featured[col].isnull().sum() > 0:

        median_value = df_featured[col].median()

        df_featured[col] = df_featured[col].fillna(
            median_value
        )


print("\nDataset shape after feature engineering:")
print(df_featured.shape)

print("\nTotal remaining missing values:")
print(df_featured.isnull().sum().sum())


print("\nSample of New Features:")

print(
    df_featured[new_features].head()
)


# =====================================================================
# 5. CREATE X AND y
# =====================================================================

X = df_featured.drop(
    columns=["TARGET"]
)

y = df_featured["TARGET"]


print("\n" + "=" * 80)
print("X AND y")
print("=" * 80)

print("X Shape:", X.shape)
print("y Shape:", y.shape)


print("\nTarget Counts:")

print(
    y.value_counts()
)


print("\nTarget Percentage:")

print(
    y.value_counts(normalize=True) * 100
)


# =====================================================================
# 6. TRAIN TEST SPLIT
# =====================================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)


print("\n" + "=" * 80)
print("TRAIN TEST SPLIT")
print("=" * 80)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("y_train:", y_train.shape)
print("y_test :", y_test.shape)


# =====================================================================
# 7. DECISION TREE CLASSIFIER
# =====================================================================

print("\n" + "=" * 80)
print("DECISION TREE CLASSIFIER")
print("=" * 80)


dt_model = DecisionTreeClassifier(

    max_depth=10,

    min_samples_split=10,

    min_samples_leaf=5,

    random_state=42
)


# Train Decision Tree

dt_model.fit(
    X_train,
    y_train
)


# Predictions

dt_train_pred = dt_model.predict(X_train)

dt_test_pred = dt_model.predict(X_test)


# Probability predictions

dt_probability = dt_model.predict_proba(
    X_test
)[:, 1]


# Confusion Matrix

dt_cm = confusion_matrix(
    y_test,
    dt_test_pred
)

dt_tn, dt_fp, dt_fn, dt_tp = dt_cm.ravel()


# Metrics

dt_train_accuracy = accuracy_score(
    y_train,
    dt_train_pred
)

dt_accuracy = accuracy_score(
    y_test,
    dt_test_pred
)

dt_precision = precision_score(
    y_test,
    dt_test_pred,
    zero_division=0
)

dt_recall = recall_score(
    y_test,
    dt_test_pred,
    zero_division=0
)

dt_specificity = (
    dt_tn /
    (dt_tn + dt_fp)
)

dt_f1 = f1_score(
    y_test,
    dt_test_pred,
    zero_division=0
)

dt_auc = roc_auc_score(
    y_test,
    dt_probability
)


# Print Decision Tree Results

print("Training Accuracy :", round(dt_train_accuracy, 4))
print("Testing Accuracy  :", round(dt_accuracy, 4))

print("Precision         :", round(dt_precision, 4))
print("Recall            :", round(dt_recall, 4))
print("Specificity       :", round(dt_specificity, 4))
print("F1 Score          :", round(dt_f1, 4))
print("ROC-AUC           :", round(dt_auc, 4))


print("\nConfusion Matrix:")

print(dt_cm)


print("\nTrue Negative  :", dt_tn)
print("False Positive :", dt_fp)
print("False Negative :", dt_fn)
print("True Positive  :", dt_tp)


print("\nClassification Report:")

print(
    classification_report(
        y_test,
        dt_test_pred,
        zero_division=0
    )
)


# =====================================================================
# 8. RANDOM FOREST CLASSIFIER
# =====================================================================

print("\n" + "=" * 80)
print("RANDOM FOREST CLASSIFIER")
print("=" * 80)


rf_model = RandomForestClassifier(

    n_estimators=100,

    max_depth=15,

    min_samples_split=10,

    min_samples_leaf=5,

    random_state=42,

    n_jobs=-1
)


# Train Random Forest

rf_model.fit(
    X_train,
    y_train
)


# Predictions

rf_train_pred = rf_model.predict(
    X_train
)

rf_test_pred = rf_model.predict(
    X_test
)


# Probability

rf_probability = rf_model.predict_proba(
    X_test
)[:, 1]


# Confusion Matrix

rf_cm = confusion_matrix(
    y_test,
    rf_test_pred
)

rf_tn, rf_fp, rf_fn, rf_tp = rf_cm.ravel()


# Metrics

rf_train_accuracy = accuracy_score(
    y_train,
    rf_train_pred
)

rf_accuracy = accuracy_score(
    y_test,
    rf_test_pred
)

rf_precision = precision_score(
    y_test,
    rf_test_pred,
    zero_division=0
)

rf_recall = recall_score(
    y_test,
    rf_test_pred,
    zero_division=0
)

rf_specificity = (
    rf_tn /
    (rf_tn + rf_fp)
)

rf_f1 = f1_score(
    y_test,
    rf_test_pred,
    zero_division=0
)

rf_auc = roc_auc_score(
    y_test,
    rf_probability
)


# Print Random Forest Results

print("Training Accuracy :", round(rf_train_accuracy, 4))
print("Testing Accuracy  :", round(rf_accuracy, 4))

print("Precision         :", round(rf_precision, 4))
print("Recall            :", round(rf_recall, 4))
print("Specificity       :", round(rf_specificity, 4))
print("F1 Score          :", round(rf_f1, 4))
print("ROC-AUC           :", round(rf_auc, 4))


print("\nConfusion Matrix:")

print(rf_cm)


print("\nTrue Negative  :", rf_tn)
print("False Positive :", rf_fp)
print("False Negative :", rf_fn)
print("True Positive  :", rf_tp)


print("\nClassification Report:")

print(
    classification_report(
        y_test,
        rf_test_pred,
        zero_division=0
    )
)


# =====================================================================
# 9. BAGGING CLASSIFIER
# =====================================================================

print("\n" + "=" * 80)
print("BAGGING CLASSIFIER")
print("=" * 80)


bag_model = BaggingClassifier(

    estimator=DecisionTreeClassifier(

        max_depth=10,

        min_samples_leaf=5,

        random_state=42
    ),

    n_estimators=50,

    random_state=42,

    n_jobs=-1
)


# Train Bagging

bag_model.fit(
    X_train,
    y_train
)


# Predictions

bag_train_pred = bag_model.predict(
    X_train
)

bag_test_pred = bag_model.predict(
    X_test
)


# Probability

bag_probability = bag_model.predict_proba(
    X_test
)[:, 1]


# Confusion Matrix

bag_cm = confusion_matrix(
    y_test,
    bag_test_pred
)

bag_tn, bag_fp, bag_fn, bag_tp = bag_cm.ravel()


# Metrics

bag_train_accuracy = accuracy_score(
    y_train,
    bag_train_pred
)

bag_accuracy = accuracy_score(
    y_test,
    bag_test_pred
)

bag_precision = precision_score(
    y_test,
    bag_test_pred,
    zero_division=0
)

bag_recall = recall_score(
    y_test,
    bag_test_pred,
    zero_division=0
)

bag_specificity = (
    bag_tn /
    (bag_tn + bag_fp)
)

bag_f1 = f1_score(
    y_test,
    bag_test_pred,
    zero_division=0
)

bag_auc = roc_auc_score(
    y_test,
    bag_probability
)


# Print Bagging Results

print("Training Accuracy :", round(bag_train_accuracy, 4))
print("Testing Accuracy  :", round(bag_accuracy, 4))

print("Precision         :", round(bag_precision, 4))
print("Recall            :", round(bag_recall, 4))
print("Specificity       :", round(bag_specificity, 4))
print("F1 Score          :", round(bag_f1, 4))
print("ROC-AUC           :", round(bag_auc, 4))


print("\nConfusion Matrix:")

print(bag_cm)


print("\nTrue Negative  :", bag_tn)
print("False Positive :", bag_fp)
print("False Negative :", bag_fn)
print("True Positive  :", bag_tp)


print("\nClassification Report:")

print(
    classification_report(
        y_test,
        bag_test_pred,
        zero_division=0
    )
)


# =====================================================================
# 10. FINAL MODEL COMPARISON
# =====================================================================

results = pd.DataFrame({

    "Model": [
        "Decision Tree",
        "Random Forest",
        "Bagging"
    ],

    "Train Accuracy": [
        dt_train_accuracy,
        rf_train_accuracy,
        bag_train_accuracy
    ],

    "Test Accuracy": [
        dt_accuracy,
        rf_accuracy,
        bag_accuracy
    ],

    "Precision": [
        dt_precision,
        rf_precision,
        bag_precision
    ],

    "Recall": [
        dt_recall,
        rf_recall,
        bag_recall
    ],

    "Specificity": [
        dt_specificity,
        rf_specificity,
        bag_specificity
    ],

    "F1 Score": [
        dt_f1,
        rf_f1,
        bag_f1
    ],

    "ROC-AUC": [
        dt_auc,
        rf_auc,
        bag_auc
    ]
})


# Sort according to ROC-AUC

results = results.sort_values(
    by="ROC-AUC",
    ascending=False
)


print("\n" + "=" * 110)
print("FINAL MODEL COMPARISON AFTER ADDING 20 FEATURES")
print("=" * 110)

print(
    results.round(4).to_string(
        index=False
    )
)


# =====================================================================
# 11. FIND BEST MODEL
# =====================================================================

best_model = results.iloc[0]


print("\n" + "=" * 80)
print("BEST MODEL")
print("=" * 80)

print(
    "Model       :",
    best_model["Model"]
)

print(
    "Accuracy    :",
    round(best_model["Test Accuracy"], 4)
)

print(
    "Precision   :",
    round(best_model["Precision"], 4)
)

print(
```

```python
# =============================================================================
# HOME CREDIT - COMPLETE FEATURE ENGINEERING
# =============================================================================

import pandas as pd
import numpy as np


# =============================================================================
# CREATE COPY OF CLEANED DATASET
# =============================================================================

df_featured = df.copy()

# Small number to prevent division by zero
EPS = 1e-6


print("=" * 80)
print("HOME CREDIT FEATURE ENGINEERING")
print("=" * 80)

print("Dataset Shape Before Feature Engineering:", df_featured.shape)


# =============================================================================
# FEATURE 1: CREDIT TO INCOME RATIO
# =============================================================================
# Measures the size of the requested loan compared with annual income.
#
# Higher value = larger loan burden relative to income.

df_featured["CREDIT_INCOME_RATIO"] = (
    df_featured["AMT_CREDIT"]
    /
    (df_featured["AMT_INCOME_TOTAL"] + EPS)
)


# =============================================================================
# FEATURE 2: ANNUITY TO INCOME RATIO
# =============================================================================
# Measures loan payment burden relative to applicant income.

df_featured["ANNUITY_INCOME_RATIO"] = (
    df_featured["AMT_ANNUITY"]
    /
    (df_featured["AMT_INCOME_TOTAL"] + EPS)
)


# =============================================================================
# FEATURE 3: CREDIT TO ANNUITY RATIO
# =============================================================================
# Approximate indicator of how large the loan is relative to periodic payments.

df_featured["CREDIT_ANNUITY_RATIO"] = (
    df_featured["AMT_CREDIT"]
    /
    (df_featured["AMT_ANNUITY"] + EPS)
)


# =============================================================================
# FEATURE 4: GOODS PRICE TO CREDIT RATIO
# =============================================================================
# Compares value of purchased goods/property with total loan amount.

df_featured["GOODS_CREDIT_RATIO"] = (
    df_featured["AMT_GOODS_PRICE"]
    /
    (df_featured["AMT_CREDIT"] + EPS)
)


# =============================================================================
# FEATURE 5: CREDIT - GOODS PRICE DIFFERENCE
# =============================================================================

df_featured["CREDIT_GOODS_DIFF"] = (
    df_featured["AMT_CREDIT"]
    -
    df_featured["AMT_GOODS_PRICE"]
)


# =============================================================================
# FEATURE 6: CREDIT - INCOME DIFFERENCE
# =============================================================================

df_featured["CREDIT_INCOME_DIFF"] = (
    df_featured["AMT_CREDIT"]
    -
    df_featured["AMT_INCOME_TOTAL"]
)


# =============================================================================
# FEATURE 7: GOODS PRICE TO INCOME RATIO
# =============================================================================

df_featured["GOODS_INCOME_RATIO"] = (
    df_featured["AMT_GOODS_PRICE"]
    /
    (df_featured["AMT_INCOME_TOTAL"] + EPS)
)


# =============================================================================
# FEATURE 8: INCOME PER FAMILY MEMBER
# =============================================================================
# Household income adjusted for family size.

df_featured["INCOME_PER_PERSON"] = (
    df_featured["AMT_INCOME_TOTAL"]
    /
    (df_featured["CNT_FAM_MEMBERS"] + EPS)
)


# =============================================================================
# FEATURE 9: CREDIT PER FAMILY MEMBER
# =============================================================================

df_featured["CREDIT_PER_PERSON"] = (
    df_featured["AMT_CREDIT"]
    /
    (df_featured["CNT_FAM_MEMBERS"] + EPS)
)


# =============================================================================
# FEATURE 10: ANNUITY PER FAMILY MEMBER
# =============================================================================

df_featured["ANNUITY_PER_PERSON"] = (
    df_featured["AMT_ANNUITY"]
    /
    (df_featured["CNT_FAM_MEMBERS"] + EPS)
)


# =============================================================================
# FEATURE 11: GOODS PRICE PER FAMILY MEMBER
# =============================================================================

df_featured["GOODS_PER_PERSON"] = (
    df_featured["AMT_GOODS_PRICE"]
    /
    (df_featured["CNT_FAM_MEMBERS"] + EPS)
)


# =============================================================================
# FEATURE 12: CHILDREN RATIO
# =============================================================================
# Percentage/proportion of family members who are children.

df_featured["CHILDREN_RATIO"] = (
    df_featured["CNT_CHILDREN"]
    /
    (df_featured["CNT_FAM_MEMBERS"] + EPS)
)


# =============================================================================
# FEATURE 13: ADULT FAMILY MEMBERS
# =============================================================================

df_featured["ADULT_COUNT"] = (
    df_featured["CNT_FAM_MEMBERS"]
    -
    df_featured["CNT_CHILDREN"]
)


# =============================================================================
# FEATURE 14: INCOME PER CHILD
# =============================================================================

df_featured["INCOME_PER_CHILD"] = (
    df_featured["AMT_INCOME_TOTAL"]
    /
    (df_featured["CNT_CHILDREN"] + 1)
)


# =============================================================================
# FEATURE 15: AGE IN YEARS
# =============================================================================
# DAYS_BIRTH is represented as negative days.

df_featured["AGE_YEARS"] = (
    -df_featured["DAYS_BIRTH"]
    / 365.25
)


# =============================================================================
# FEATURE 16: EMPLOYMENT YEARS
# =============================================================================
# 365243 is a special/sentinel value in DAYS_EMPLOYED.
# Replace it before calculating employment duration.

employment_days = df_featured["DAYS_EMPLOYED"].copy()

employment_days = employment_days.where(
    employment_days.abs() < 365243,
    np.nan
)

df_featured["EMPLOYMENT_YEARS"] = (
    -employment_days
    / 365.25
)


# =============================================================================
# FEATURE 17: EMPLOYMENT TO AGE RATIO
# =============================================================================
# Measures how much of the applicant's life is represented by employment.

df_featured["EMPLOYMENT_AGE_RATIO"] = (
    df_featured["EMPLOYMENT_YEARS"]
    /
    (df_featured["AGE_YEARS"] + EPS)
)


# =============================================================================
# FEATURE 18: AGE WHEN EMPLOYMENT STARTED
# =============================================================================

df_featured["AGE_AT_EMPLOYMENT_START"] = (
    df_featured["AGE_YEARS"]
    -
    df_featured["EMPLOYMENT_YEARS"]
)


# =============================================================================
# FEATURE 19: REGISTRATION YEARS
# =============================================================================

df_featured["REGISTRATION_YEARS"] = (
    -df_featured["DAYS_REGISTRATION"]
    / 365.25
)


# =============================================================================
# FEATURE 20: ID PUBLISH YEARS
# =============================================================================

df_featured["ID_PUBLISH_YEARS"] = (
    -df_featured["DAYS_ID_PUBLISH"]
    / 365.25
)


# =============================================================================
# FEATURE 21: PHONE CHANGE YEARS
# =============================================================================

if "DAYS_LAST_PHONE_CHANGE" in df_featured.columns:

    df_featured["PHONE_CHANGE_YEARS"] = (
        -df_featured["DAYS_LAST_PHONE_CHANGE"]
        / 365.25
    )


# =============================================================================
# FEATURE 22: CAR AGE
# =============================================================================

if "OWN_CAR_AGE" in df_featured.columns:

    df_featured["CAR_AGE_RATIO"] = (
        df_featured["OWN_CAR_AGE"]
        /
        (df_featured["AGE_YEARS"] + EPS)
    )


# =============================================================================
# FEATURE 23: EXTERNAL SOURCE MEAN
# =============================================================================

ext_columns = [
    col
    for col in [
        "EXT_SOURCE_1",
        "EXT_SOURCE_2",
        "EXT_SOURCE_3"
    ]
    if col in df_featured.columns
]

if len(ext_columns) > 0:

    df_featured["EXT_SOURCE_MEAN"] = (
        df_featured[ext_columns]
        .mean(axis=1)
    )


# =============================================================================
# FEATURE 24: EXTERNAL SOURCE MINIMUM
# =============================================================================

if len(ext_columns) > 0:

    df_featured["EXT_SOURCE_MIN"] = (
        df_featured[ext_columns]
        .min(axis=1)
    )


# =============================================================================
# FEATURE 25: EXTERNAL SOURCE MAXIMUM
# =============================================================================

if len(ext_columns) > 0:

    df_featured["EXT_SOURCE_MAX"] = (
        df_featured[ext_columns]
        .max(axis=1)
    )


# =============================================================================
# FEATURE 26: EXTERNAL SOURCE STANDARD DEVIATION
# =============================================================================

if len(ext_columns) > 1:

    df_featured["EXT_SOURCE_STD"] = (
        df_featured[ext_columns]
        .std(axis=1)
    )


# =============================================================================
# FEATURE 27: EXTERNAL SOURCE RANGE
# =============================================================================

if len(ext_columns) > 0:

    df_featured["EXT_SOURCE_RANGE"] = (
        df_featured["EXT_SOURCE_MAX"]
        -
        df_featured["EXT_SOURCE_MIN"]
    )


# =============================================================================
# FEATURE 28: EXTERNAL SOURCE PRODUCT
# =============================================================================

if len(ext_columns) == 3:

    df_featured["EXT_SOURCE_PRODUCT"] = (
        df_featured["EXT_SOURCE_1"]
        *
        df_featured["EXT_SOURCE_2"]
        *
        df_featured["EXT_SOURCE_3"]
    )


# =============================================================================
# FEATURE 29: EXT SOURCE 2 AND 3 AVERAGE
# =============================================================================

if (
    "EXT_SOURCE_2" in df_featured.columns
    and
    "EXT_SOURCE_3" in df_featured.columns
):

    df_featured["EXT_SOURCE_2_3_MEAN"] = (
        df_featured["EXT_SOURCE_2"]
        +
        df_featured["EXT_SOURCE_3"]
    ) / 2


# =============================================================================
# FEATURE 30: 30-DAY SOCIAL DEFAULT RATIO
# =============================================================================

if (
    "DEF_30_CNT_SOCIAL_CIRCLE" in df_featured.columns
    and
    "OBS_30_CNT_SOCIAL_CIRCLE" in df_featured.columns
):

    df_featured["SOCIAL_DEFAULT_30_RATIO"] = (
        df_featured["DEF_30_CNT_SOCIAL_CIRCLE"]
        /
        (
            df_featured["OBS_30_CNT_SOCIAL_CIRCLE"]
            + EPS
        )
    )


# =============================================================================
# FEATURE 31: 60-DAY SOCIAL DEFAULT RATIO
# =============================================================================

if (
    "DEF_60_CNT_SOCIAL_CIRCLE" in df_featured.columns
    and
    "OBS_60_CNT_SOCIAL_CIRCLE" in df_featured.columns
):

    df_featured["SOCIAL_DEFAULT_60_RATIO"] = (
        df_featured["DEF_60_CNT_SOCIAL_CIRCLE"]
        /
        (
            df_featured["OBS_60_CNT_SOCIAL_CIRCLE"]
            + EPS
        )
    )


# =============================================================================
# FEATURE 32: TOTAL SOCIAL OBSERVATIONS
# =============================================================================

if (
    "OBS_30_CNT_SOCIAL_CIRCLE" in df_featured.columns
    and
    "OBS_60_CNT_SOCIAL_CIRCLE" in df_featured.columns
):

    df_featured["TOTAL_SOCIAL_OBS"] = (
        df_featured["OBS_30_CNT_SOCIAL_CIRCLE"]
        +
        df_featured["OBS_60_CNT_SOCIAL_CIRCLE"]
    )


# =============================================================================
# FEATURE 33: TOTAL SOCIAL DEFAULTS
# =============================================================================

if (
    "DEF_30_CNT_SOCIAL_CIRCLE" in df_featured.columns
    and
    "DEF_60_CNT_SOCIAL_CIRCLE" in df_featured.columns
):

    df_featured["TOTAL_SOCIAL_DEFAULTS"] = (
        df_featured["DEF_30_CNT_SOCIAL_CIRCLE"]
        +
        df_featured["DEF_60_CNT_SOCIAL_CIRCLE"]
    )


# =============================================================================
# FEATURE 34: TOTAL DOCUMENTS SUBMITTED
# =============================================================================

document_columns = [
    col
    for col in df_featured.columns
    if col.startswith("FLAG_DOCUMENT_")
]

if len(document_columns) > 0:

    df_featured["TOTAL_DOCUMENTS"] = (
        df_featured[document_columns]
        .sum(axis=1)
    )


# =============================================================================
# FEATURE 35: TOTAL CONTACT INFORMATION FLAGS
# =============================================================================

contact_columns = [
    col
    for col in [
        "FLAG_MOBIL",
        "FLAG_EMP_PHONE",
        "FLAG_WORK_PHONE",
        "FLAG_CONT_MOBILE",
        "FLAG_PHONE",
        "FLAG_EMAIL"
    ]
    if col in df_featured.columns
]

if len(contact_columns) > 0:

    df_featured["TOTAL_CONTACT_FLAGS"] = (
        df_featured[contact_columns]
        .sum(axis=1)
    )


# =============================================================================
# FEATURE 36: REGION RATING AVERAGE
# =============================================================================

if (
    "REGION_RATING_CLIENT" in df_featured.columns
    and
    "REGION_RATING_CLIENT_W_CITY" in df_featured.columns
):

    df_featured["REGION_RATING_MEAN"] = (
        df_featured["REGION_RATING_CLIENT"]
        +
        df_featured["REGION_RATING_CLIENT_W_CITY"]
    ) / 2


# =============================================================================
# FEATURE 37: REGION RATING DIFFERENCE
# =============================================================================

if (
    "REGION_RATING_CLIENT" in df_featured.columns
    and
    "REGION_RATING_CLIENT_W_CITY" in df_featured.columns
):

    df_featured["REGION_RATING_DIFF"] = (
        df_featured["REGION_RATING_CLIENT_W_CITY"]
        -
        df_featured["REGION_RATING_CLIENT"]
    )


# =============================================================================
# FEATURE 38: HOUR OF APPLICATION
# =============================================================================
# Convert application hour into broader time periods.

if "HOUR_APPR_PROCESS_START" in df_featured.columns:

    df_featured["APPLICATION_TIME_PERIOD"] = pd.cut(
        df_featured["HOUR_APPR_PROCESS_START"],
        bins=[-1, 5, 11, 17, 23],
        labels=[0, 1, 2, 3]
    )

    df_featured["APPLICATION_TIME_PERIOD"] = (
        df_featured["APPLICATION_TIME_PERIOD"]
        .astype(int)
    )


# =============================================================================
# FEATURE 39: CREDIT AMOUNT PER YEAR OF AGE
# =============================================================================

df_featured["CREDIT_PER_AGE"] = (
    df_featured["AMT_CREDIT"]
    /
    (df_featured["AGE_YEARS"] + EPS)
)


# =============================================================================
# FEATURE 40: INCOME PER YEAR OF AGE
# =============================================================================

df_featured["INCOME_PER_AGE"] = (
    df_featured["AMT_INCOME_TOTAL"]
    /
    (df_featured["AGE_YEARS"] + EPS)
)


# =============================================================================
# FEATURE 41: ANNUITY PER YEAR OF AGE
# =============================================================================

df_featured["ANNUITY_PER_AGE"] = (
    df_featured["AMT_ANNUITY"]
    /
    (df_featured["AGE_YEARS"] + EPS)
)


# =============================================================================
# FEATURE 42: CREDIT RELATIVE TO EMPLOYMENT YEARS
# =============================================================================

df_featured["CREDIT_EMPLOYMENT_RATIO"] = (
    df_featured["AMT_CREDIT"]
    /
    (
        df_featured["EMPLOYMENT_YEARS"]
        + 1
    )
)


# =============================================================================
# FEATURE 43: INCOME RELATIVE TO EMPLOYMENT YEARS
# =============================================================================

df_featured["INCOME_EMPLOYMENT_RATIO"] = (
    df_featured["AMT_INCOME_TOTAL"]
    /
    (
        df_featured["EMPLOYMENT_YEARS"]
        + 1
    )
)


# =============================================================================
# FEATURE 44: ANNUITY / CREDIT
# =============================================================================

df_featured["ANNUITY_CREDIT_RATIO"] = (
    df_featured["AMT_ANNUITY"]
    /
    (df_featured["AMT_CREDIT"] + EPS)
)


# =============================================================================
# FEATURE 45: CREDIT REMAINING AFTER ONE ANNUITY
# =============================================================================

df_featured["CREDIT_AFTER_ANNUITY"] = (
    df_featured["AMT_CREDIT"]
    -
    df_featured["AMT_ANNUITY"]
)


# =============================================================================
# CLEAN INFINITE VALUES
# =============================================================================

df_featured.replace(
    [np.inf, -np.inf],
    np.nan,
    inplace=True
)


# =============================================================================
# FILL NULL VALUES CREATED BY FEATURE ENGINEERING
# =============================================================================

new_columns = [
    col
    for col in df_featured.columns
    if col not in df.columns
]


for col in new_columns:

    if df_featured[col].isnull().sum() > 0:

        median_value = (
            df_featured[col]
            .median()
        )

        df_featured[col] = (
            df_featured[col]
            .fillna(median_value)
        )


# =============================================================================
# CONVERT NEW FLOAT64 FEATURES TO FLOAT32
# =============================================================================

for col in new_columns:

    if df_featured[col].dtype == "float64":

        df_featured[col] = (
            df_featured[col]
            .astype("float32")
        )


# =============================================================================
# FINAL FEATURE ENGINEERING REPORT
# =============================================================================

print("\n" + "=" * 80)
print("FEATURE ENGINEERING COMPLETED")
print("=" * 80)


print(
    "Original Number of Columns :",
    df.shape[1]
)


print(
    "Number of New Features     :",
    len(new_columns)
)


print(
    "Final Number of Columns    :",
    df_featured.shape[1]
)


print(
    "Number of Rows             :",
    df_featured.shape[0]
)


print(
    "Final Dataset Shape        :",
    df_featured.shape
)


print(
    "Remaining Missing Values   :",
    df_featured.isnull().sum().sum()
)


print(
    "Remaining Infinite Values  :",
    np.isinf(
        df_featured.select_dtypes(
            include=np.number
        )
    ).sum().sum()
)


# =============================================================================
# PRINT ALL NEW FEATURES
# =============================================================================

print("\n" + "=" * 80)
print("NEWLY CREATED FEATURES")
print("=" * 80)


for number, column in enumerate(
    new_columns,
    start=1
):

    print(
        number,
        column
    )


# =============================================================================
# DISPLAY NEW FEATURES
# =============================================================================

print("\n" + "=" * 80)
print("SAMPLE OF ENGINEERED FEATURES")
print("=" * 80)


print(
    df_featured[
        new_columns
    ].head()
)


# =============================================================================
# FINAL DATAFRAME
# =============================================================================
# Use df_featured for model training.
#
# Example:
#
# X = df_featured.drop("TARGET", axis=1)
# y = df_featured["TARGET"]
#
# =============================================================================
```

```python
# =============================================================================
# HOME CREDIT DEFAULT RISK
# COMPLETE END-TO-END MODEL BUILDING
#
# MODELS:
# 1. DECISION TREE CLASSIFIER
# 2. RANDOM FOREST CLASSIFIER
# 3. BAGGING CLASSIFIER
#
# HYPERPARAMETER TUNING:
# GridSearchCV
#
# CROSS VALIDATION:
# 5-Fold Stratified Cross Validation
#
# PRIMARY MODEL-SELECTION METRIC:
# ROC-AUC
#
# INPUT DATAFRAME:
# df_featured
#
# TARGET:
# TARGET
# =============================================================================


# =============================================================================
# 1. IMPORT LIBRARIES
# =============================================================================

import pandas as pd
import numpy as np
import time
import warnings
import joblib

warnings.filterwarnings("ignore")


from sklearn.model_selection import (
    train_test_split,
    GridSearchCV,
    StratifiedKFold,
    cross_val_score
)


from sklearn.tree import (
    DecisionTreeClassifier
)


from sklearn.ensemble import (
    RandomForestClassifier,
    BaggingClassifier
)


from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)


# =============================================================================
# 2. CHECK DATASET
# =============================================================================

print("=" * 120)
print("HOME CREDIT DEFAULT RISK - COMPLETE GRID SEARCH PIPELINE")
print("=" * 120)


print(
    "Dataset Shape:",
    df_featured.shape
)


print(
    "Number of Rows:",
    df_featured.shape[0]
)


print(
    "Number of Columns:",
    df_featured.shape[1]
)


# =============================================================================
# 3. CHECK TARGET COLUMN
# =============================================================================

if "TARGET" not in df_featured.columns:

    raise ValueError(
        "TARGET column is missing from df_featured."
    )


print(
    "\nTARGET column found successfully."
)


# =============================================================================
# 4. CREATE X AND y
# =============================================================================

X = df_featured.drop(
    columns=["TARGET"]
)


y = df_featured[
    "TARGET"
]


print("\n" + "=" * 120)
print("CREATE X AND y")
print("=" * 120)


print(
    "X Shape:",
    X.shape
)


print(
    "y Shape:",
    y.shape
)


# =============================================================================
# 5. TARGET DISTRIBUTION
# =============================================================================

print("\nTarget Counts:")


print(
    y.value_counts()
)


print("\nTarget Percentage:")


print(
    (
        y.value_counts(
            normalize=True
        )
        * 100
    ).round(2)
)


# =============================================================================
# 6. CHECK TARGET CLASSES
# =============================================================================

print(
    "\nTarget Classes:",
    sorted(
        y.unique()
    )
)


# =============================================================================
# 7. REPLACE INFINITE VALUES
# =============================================================================

X = X.replace(
    [
        np.inf,
        -np.inf
    ],
    np.nan
)


# =============================================================================
# 8. HANDLE REMAINING MISSING VALUES
# =============================================================================

print("\n" + "=" * 120)
print("FINAL MISSING VALUE CHECK")
print("=" * 120)


columns_with_nulls = (
    X.columns[
        X.isnull().any()
    ]
    .tolist()
)


print(
    "Columns With Missing Values:",
    len(columns_with_nulls)
)


for col in columns_with_nulls:

    median_value = (
        X[col]
        .median()
    )

    X[col] = (
        X[col]
        .fillna(
            median_value
        )
    )


print(
    "Remaining Missing Values:",
    X.isnull().sum().sum()
)


# =============================================================================
# 9. CHECK INFINITE VALUES
# =============================================================================

infinite_count = (
    np.isinf(
        X.select_dtypes(
            include=np.number
        )
    )
    .sum()
    .sum()
)


print(
    "Remaining Infinite Values:",
    infinite_count
)


# =============================================================================
# 10. VERIFY ALL COLUMNS ARE NUMERIC
# =============================================================================

non_numeric_columns = (
    X.select_dtypes(
        exclude=np.number
    )
    .columns
    .tolist()
)


print(
    "\nNon-Numeric Columns:",
    non_numeric_columns
)


if len(non_numeric_columns) > 0:

    raise ValueError(
        "Non-numeric columns are still present. "
        "Encode them before model training."
    )


print(
    "All independent variables are numeric."
)


# =============================================================================
# 11. TRAIN TEST SPLIT
# =============================================================================
#
# 80% TRAINING
# 20% TESTING
#
# stratify=y maintains the class distribution.
#
# =============================================================================

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)


print("\n" + "=" * 120)
print("TRAIN TEST SPLIT")
print("=" * 120)


print(
    "X_train Shape:",
    X_train.shape
)


print(
    "X_test Shape:",
    X_test.shape
)


print(
    "y_train Shape:",
    y_train.shape
)


print(
    "y_test Shape:",
    y_test.shape
)


# =============================================================================
# 12. TRAINING TARGET DISTRIBUTION
# =============================================================================

print("\nTraining Target Distribution:")


print(
    (
        y_train.value_counts(
            normalize=True
        )
        * 100
    ).round(2)
)


# =============================================================================
# 13. TESTING TARGET DISTRIBUTION
# =============================================================================

print("\nTesting Target Distribution:")


print(
    (
        y_test.value_counts(
            normalize=True
        )
        * 100
    ).round(2)
)


# =============================================================================
# 14. CREATE 5-FOLD STRATIFIED CROSS VALIDATION
# =============================================================================

cv = StratifiedKFold(

    n_splits=5,

    shuffle=True,

    random_state=42
)


print("\n" + "=" * 120)
print("CROSS VALIDATION CONFIGURATION")
print("=" * 120)


print(
    "Cross Validation Type : StratifiedKFold"
)


print(
    "Number of Folds       : 5"
)


print(
    "Shuffle               : True"
)


print(
    "Random State          : 42"
)


print(
    "Optimization Metric   : ROC-AUC"
)


# =============================================================================
# =============================================================================
#
# DECISION TREE
#
# =============================================================================
# =============================================================================


# =============================================================================
# 15. CREATE DECISION TREE MODEL
# =============================================================================

decision_tree = DecisionTreeClassifier(

    random_state=42
)


# =============================================================================
# 16. DECISION TREE PARAMETER GRID
# =============================================================================

dt_param_grid = {

    "criterion": [
        "gini",
        "entropy"
    ],

    "max_depth": [
        5,
        10,
        15
    ],

    "min_samples_split": [
        10,
        20
    ],

    "min_samples_leaf": [
        5,
        10
    ],

    "class_weight": [
        None,
        "balanced"
    ]
}


print("\n" + "=" * 120)
print("DECISION TREE GRID SEARCH")
print("=" * 120)


print(
    "Parameter Grid:"
)


print(
    dt_param_grid
)


# =============================================================================
# 17. DECISION TREE GRID SEARCH CV
# =============================================================================

dt_grid = GridSearchCV(

    estimator=decision_tree,

    param_grid=dt_param_grid,

    scoring="roc_auc",

    cv=cv,

    n_jobs=-1,

    verbose=1,

    return_train_score=True,

    refit=True
)


dt_start_time = (
    time.time()
)


dt_grid.fit(

    X_train,

    y_train
)


dt_grid_time = (

    time.time()

    -

    dt_start_time
)


# =============================================================================
# 18. BEST DECISION TREE
# =============================================================================

best_dt = (
    dt_grid.best_estimator_
)


print("\nBest Decision Tree Parameters:")


print(
    dt_grid.best_params_
)


print(
    "\nBest Decision Tree CV ROC-AUC:",
    round(
        dt_grid.best_score_,
        6
    )
)


print(
    "Grid Search Time:",
    round(
        dt_grid_time,
        2
    ),
    "seconds"
)


# =============================================================================
# 19. DECISION TREE PREDICTIONS
# =============================================================================

dt_train_prediction = (
    best_dt.predict(
        X_train
    )
)


dt_test_prediction = (
    best_dt.predict(
        X_test
    )
)


dt_test_probability = (
    best_dt.predict_proba(
        X_test
    )[:, 1]
)


# =============================================================================
# 20. DECISION TREE METRICS
# =============================================================================

dt_train_accuracy = (
    accuracy_score(
        y_train,
        dt_train_prediction
    )
)


dt_accuracy = (
    accuracy_score(
        y_test,
        dt_test_prediction
    )
)


dt_balanced_accuracy = (
    balanced_accuracy_score(
        y_test,
        dt_test_prediction
    )
)


dt_precision = (
    precision_score(
        y_test,
        dt_test_prediction,
        zero_division=0
    )
)


dt_recall = (
    recall_score(
        y_test,
        dt_test_prediction,
        zero_division=0
    )
)


dt_f1 = (
    f1_score(
        y_test,
        dt_test_prediction,
        zero_division=0
    )
)


dt_auc = (
    roc_auc_score(
        y_test,
        dt_test_probability
    )
)


# =============================================================================
# 21. DECISION TREE CONFUSION MATRIX
# =============================================================================

dt_confusion_matrix = (
    confusion_matrix(
        y_test,
        dt_test_prediction
    )
)


dt_tn, dt_fp, dt_fn, dt_tp = (
    dt_confusion_matrix.ravel()
)


dt_specificity = (

    dt_tn

    /

    (
        dt_tn
        +
        dt_fp
    )
)


# =============================================================================
# 22. PRINT DECISION TREE RESULTS
# =============================================================================

print("\n" + "=" * 120)
print("BEST DECISION TREE - FINAL RESULTS")
print("=" * 120)


print(
    "Training Accuracy :",
    round(
        dt_train_accuracy,
        6
    )
)


print(
    "Testing Accuracy  :",
    round(
        dt_accuracy,
        6
    )
)


print(
    "Balanced Accuracy :",
    round(
        dt_balanced_accuracy,
        6
    )
)


print(
    "Precision         :",
    round(
        dt_precision,
        6
    )
)


print(
    "Recall            :",
    round(
        dt_recall,
        6
    )
)


print(
    "Specificity       :",
    round(
        dt_specificity,
        6
    )
)


print(
    "F1 Score          :",
    round(
        dt_f1,
        6
    )
)


print(
    "ROC-AUC           :",
    round(
        dt_auc,
        6
    )
)


print("\nConfusion Matrix:")


print(
    dt_confusion_matrix
)


print(
    "\nTrue Negative:",
    dt_tn
)


print(
    "False Positive:",
    dt_fp
)


print(
    "False Negative:",
    dt_fn
)


print(
    "True Positive:",
    dt_tp
)


print("\nClassification Report:")


print(
    classification_report(

        y_test,

        dt_test_prediction,

        digits=4,

        zero_division=0
    )
)


# =============================================================================
# 23. DECISION TREE GRID SEARCH RESULTS
# =============================================================================

dt_cv_results = (
    pd.DataFrame(
        dt_grid.cv_results_
    )
)


dt_cv_results = (

    dt_cv_results[

        [
            "params",
            "mean_train_score",
            "mean_test_score",
            "std_test_score",
            "rank_test_score"
        ]

    ]

    .sort_values(
        by="rank_test_score"
    )
)


print("\nTop 10 Decision Tree Parameter Combinations:")


print(
    dt_cv_results
    .head(10)
    .to_string(
        index=False
    )
)


# =============================================================================
# =============================================================================
#
# RANDOM FOREST
#
# =============================================================================
# =============================================================================


# =============================================================================
# 24. CREATE RANDOM FOREST MODEL
# =============================================================================

random_forest = RandomForestClassifier(

    random_state=42,

    n_jobs=-1
)


# =============================================================================
# 25. RANDOM FOREST PARAMETER GRID
# =============================================================================

rf_param_grid = {

    "n_estimators": [
        100,
        200
    ],

    "max_depth": [
        10,
        15
    ],

    "min_samples_split": [
        10,
        20
    ],

    "min_samples_leaf": [
        5,
        10
    ],

    "max_features": [
        "sqrt"
    ],

    "class_weight": [
        None,
        "balanced"
    ]
}


print("\n" + "=" * 120)
print("RANDOM FOREST GRID SEARCH")
print("=" * 120)


print(
    "Parameter Grid:"
)


print(
    rf_param_grid
)


# =============================================================================
# 26. RANDOM FOREST GRID SEARCH
# =============================================================================

rf_grid = GridSearchCV(

    estimator=random_forest,

    param_grid=rf_param_grid,

    scoring="roc_auc",

    cv=cv,

    n_jobs=-1,

    verbose=1,

    return_train_score=True,

    refit=True
)


rf_start_time = (
    time.time()
)


rf_grid.fit(

    X_train,

    y_train
)


rf_grid_time = (

    time.time()

    -

    rf_start_time
)


# =============================================================================
# 27. BEST RANDOM FOREST
# =============================================================================

best_rf = (
    rf_grid.best_estimator_
)


print("\nBest Random Forest Parameters:")


print(
    rf_grid.best_params_
)


print(
    "\nBest Random Forest CV ROC-AUC:",
    round(
        rf_grid.best_score_,
        6
    )
)


print(
    "Grid Search Time:",
    round(
        rf_grid_time,
        2
    ),
    "seconds"
)


# =============================================================================
# 28. RANDOM FOREST PREDICTIONS
# =============================================================================

rf_train_prediction = (
    best_rf.predict(
        X_train
    )
)


rf_test_prediction = (
    best_rf.predict(
        X_test
    )
)


rf_test_probability = (
    best_rf.predict_proba(
        X_test
    )[:, 1]
)


# =============================================================================
# 29. RANDOM FOREST METRICS
# =============================================================================

rf_train_accuracy = (
    accuracy_score(
        y_train,
        rf_train_prediction
    )
)


rf_accuracy = (
    accuracy_score(
        y_test,
        rf_test_prediction
    )
)


rf_balanced_accuracy = (
    balanced_accuracy_score(
        y_test,
        rf_test_prediction
    )
)


rf_precision = (
    precision_score(
        y_test,
        rf_test_prediction,
        zero_division=0
    )
)


rf_recall = (
    recall_score(
        y_test,
        rf_test_prediction,
        zero_division=0
    )
)


rf_f1 = (
    f1_score(
        y_test,
        rf_test_prediction,
        zero_division=0
    )
)


rf_auc = (
    roc_auc_score(
        y_test,
        rf_test_probability
    )
)


# =============================================================================
# 30. RANDOM FOREST CONFUSION MATRIX
# =============================================================================

rf_confusion_matrix = (
    confusion_matrix(
        y_test,
        rf_test_prediction
    )
)


rf_tn, rf_fp, rf_fn, rf_tp = (
    rf_confusion_matrix.ravel()
)


rf_specificity = (

    rf_tn

    /

    (
        rf_tn
        +
        rf_fp
    )
)


# =============================================================================
# 31. PRINT RANDOM FOREST RESULTS
# =============================================================================

print("\n" + "=" * 120)
print("BEST RANDOM FOREST - FINAL RESULTS")
print("=" * 120)


print(
    "Training Accuracy :",
    round(
        rf_train_accuracy,
        6
    )
)


print(
    "Testing Accuracy  :",
    round(
        rf_accuracy,
        6
    )
)


print(
    "Balanced Accuracy :",
    round(
        rf_balanced_accuracy,
        6
    )
)


print(
    "Precision         :",
    round(
        rf_precision,
        6
    )
)


print(
    "Recall            :",
    round(
        rf_recall,
        6
    )
)


print(
    "Specificity       :",
    round(
        rf_specificity,
        6
    )
)


print(
    "F1 Score          :",
    round(
        rf_f1,
        6
    )
)


print(
    "ROC-AUC           :",
    round(
        rf_auc,
        6
    )
)


print("\nConfusion Matrix:")


print(
    rf_confusion_matrix
)


print(
    "\nTrue Negative:",
    rf_tn
)


print(
    "False Positive:",
    rf_fp
)


print(
    "False Negative:",
    rf_fn
)


print(
    "True Positive:",
    rf_tp
)


print("\nClassification Report:")


print(
    classification_report(

        y_test,

        rf_test_prediction,

        digits=4,

        zero_division=0
    )
)


# =============================================================================
# 32. RANDOM FOREST GRID SEARCH RESULTS
# =============================================================================

rf_cv_results = (
    pd.DataFrame(
        rf_grid.cv_results_
    )
)


rf_cv_results = (

    rf_cv_results[

        [
            "params",
            "mean_train_score",
            "mean_test_score",
            "std_test_score",
            "rank_test_score"
        ]

    ]

    .sort_values(
        by="rank_test_score"
    )
)


print("\nTop 10 Random Forest Parameter Combinations:")


print(
    rf_cv_results
    .head(10)
    .to_string(
        index=False
    )
)


# =============================================================================
# =============================================================================
#
# BAGGING CLASSIFIER
#
# =============================================================================
# =============================================================================


# =============================================================================
# 33. CREATE BASE DECISION TREE FOR BAGGING
# =============================================================================

base_tree = DecisionTreeClassifier(

    random_state=42
)


# =============================================================================
# 34. CREATE BAGGING CLASSIFIER
# =============================================================================

bagging =BaggingClassifier(

    estimator=base_tree,

    random_state=42,

    n_jobs=-1
)


# =============================================================================
# 35. BAGGING PARAMETER GRID
# =============================================================================

bag_param_grid = {

    "n_estimators": [
        25,
        50
    ],

    "max_samples": [
        0.7,
        1.0
    ],

    "max_features": [
        0.7,
        1.0
    ],

    "bootstrap": [
        True
    ],

    "estimator__max_depth": [
        5,
        10
    ],

    "estimator__min_samples_split": [
        10,
        20
    ],

    "estimator__min_samples_leaf": [
        5,
        10
    ],

    "estimator__class_weight": [
        None,
        "balanced"
    ]
}


print("\n" + "=" * 120)
print("BAGGING CLASSIFIER GRID SEARCH")
print("=" * 120)

print("Parameter Grid:")

print(
    bag_param_grid
)


# =============================================================================
# 36. BAGGING GRID SEARCH CV
# =============================================================================

bag_grid = GridSearchCV(

    estimator=bagging,

    param_grid=bag_param_grid,

    scoring="roc_auc",

    cv=cv,

    n_jobs=-1,

    verbose=1,

    return_train_score=True,

    refit=True
)


bag_start_time = (
    time.time()
)


bag_grid.fit(

    X_train,

    y_train
)


bag_grid_time = (

    time.time()

    -

    bag_start_time
)


# =============================================================================
# 37. BEST BAGGING MODEL
# =============================================================================

best_bag = (
    bag_grid.best_estimator_
)


print("\nBest Bagging Parameters:")

print(
    bag_grid.best_params_
)


print(
    "\nBest Bagging CV ROC-AUC:",
    round(
        bag_grid.best_score_,
        6
    )
)


print(
    "Grid Search Time:",
    round(
        bag_grid_time,
        2
    ),
    "seconds"
)


# =============================================================================
# 38. BAGGING PREDICTIONS
# =============================================================================

bag_train_prediction = (
    best_bag.predict(
        X_train
    )
)


bag_test_prediction = (
    best_bag.predict(
        X_test
    )
)


bag_test_probability = (
    best_bag.predict_proba(
        X_test
    )[:, 1]
)


# =============================================================================
# 39. BAGGING METRICS
# =============================================================================

bag_train_accuracy = (
    accuracy_score(
        y_train,
        bag_train_prediction
    )
)


bag_accuracy = (
    accuracy_score(
        y_test,
        bag_test_prediction
    )
)


bag_balanced_accuracy = (
    balanced_accuracy_score(
        y_test,
        bag_test_prediction
    )
)


bag_precision = (
    precision_score(
        y_test,
        bag_test_prediction,
        zero_division=0
    )
)


bag_recall = (
    recall_score(
        y_test,
        bag_test_prediction,
        zero_division=0
    )
)


bag_f1 = (
    f1_score(
        y_test,
        bag_test_prediction,
        zero_division=0
    )
)


bag_auc = (
    roc_auc_score(
        y_test,
        bag_test_probability
    )
)


# =============================================================================
# 40. BAGGING CONFUSION MATRIX
# =============================================================================

bag_confusion_matrix = (
    confusion_matrix(
        y_test,
        bag_test_prediction
    )
)


bag_tn, bag_fp, bag_fn, bag_tp = (
    bag_confusion_matrix.ravel()
)


bag_specificity = (

    bag_tn

    /

    (
        bag_tn
        +
        bag_fp
    )
)


# =============================================================================
# 41. PRINT BAGGING RESULTS
# =============================================================================

print("\n" + "=" * 120)
print("BEST BAGGING CLASSIFIER - FINAL RESULTS")
print("=" * 120)


print(
    "Training Accuracy :",
    round(
        bag_train_accuracy,
        6
    )
)


print(
    "Testing Accuracy  :",
    round(
        bag_accuracy,
        6
    )
)


print(
    "Balanced Accuracy :",
    round(
        bag_balanced_accuracy,
        6
    )
)


print(
    "Precision         :",
    round(
        bag_precision,
        6
    )
)


print(
    "Recall            :",
    round(
        bag_recall,
        6
    )
)


print(
    "Specificity       :",
    round(
        bag_specificity,
        6
    )
)


print(
    "F1 Score          :",
    round(
        bag_f1,
        6
    )
)


print(
    "ROC-AUC           :",
    round(
        bag_auc,
        6
    )
)


print("\nConfusion Matrix:")


print(
    bag_confusion_matrix
)


print(
    "\nTrue Negative:",
    bag_tn
)


print(
    "False Positive:",
    bag_fp
)


print(
    "False Negative:",
    bag_fn
)


print(
    "True Positive:",
    bag_tp
)


print("\nClassification Report:")


print(
    classification_report(

        y_test,

        bag_test_prediction,

        digits=4,

        zero_division=0
    )
)


# =============================================================================
# 42. BAGGING GRID SEARCH RESULTS
# =============================================================================

bag_cv_results = (
    pd.DataFrame(
        bag_grid.cv_results_
    )
)


bag_cv_results = (

    bag_cv_results[

        [
            "params",
            "mean_train_score",
            "mean_test_score",
            "std_test_score",
            "rank_test_score"
        ]

    ]

    .sort_values(
        by="rank_test_score"
    )
)


print("\nTop 10 Bagging Parameter Combinations:")


print(
    bag_cv_results
    .head(10)
    .to_string(
        index=False
    )
)


# =============================================================================
# 43. CREATE FINAL MODEL COMPARISON
# =============================================================================

model_results = pd.DataFrame({

    "Model": [

        "Decision Tree",

        "Random Forest",

        "Bagging Classifier"
    ],


    "Best CV ROC-AUC": [

        dt_grid.best_score_,

        rf_grid.best_score_,

        bag_grid.best_score_
    ],


    "Train Accuracy": [

        dt_train_accuracy,

        rf_train_accuracy,

        bag_train_accuracy
    ],


    "Test Accuracy": [

        dt_accuracy,

        rf_accuracy,

        bag_accuracy
    ],


    "Balanced Accuracy": [

        dt_balanced_accuracy,

        rf_balanced_accuracy,

        bag_balanced_accuracy
    ],


    "Precision": [

        dt_precision,

        rf_precision,

        bag_precision
    ],


    "Recall": [

        dt_recall,

        rf_recall,

        bag_recall
    ],


    "Specificity": [

        dt_specificity,

        rf_specificity,

        bag_specificity
    ],


    "F1 Score": [

        dt_f1,

        rf_f1,

        bag_f1
    ],


    "Test ROC-AUC": [

        dt_auc,

        rf_auc,

        bag_auc
    ],


    "Grid Search Time": [

        dt_grid_time,

        rf_grid_time,

        bag_grid_time
    ]
})


# =============================================================================
# 44. ADD OVERFITTING GAP
# =============================================================================

model_results[
    "Accuracy Gap"
] = (

    model_results[
        "Train Accuracy"
    ]

    -

    model_results[
        "Test Accuracy"
    ]
)


# =============================================================================
# 45. SORT MODELS ACCORDING TO CV ROC-AUC
# =============================================================================

model_results = (
    model_results
    .sort_values(

        by="Best CV ROC-AUC",

        ascending=False
    )
    .reset_index(
        drop=True
    )
)


# =============================================================================
# 46. PRINT FINAL COMPARISON
# =============================================================================

print("\n")
print("=" * 140)
print("FINAL MODEL COMPARISON")
print("=" * 140)


print(
    model_results
    .round(6)
    .to_string(
        index=False
    )
)


# =============================================================================
# 47. PRINT BEST PARAMETERS OF ALL MODELS
# =============================================================================

print("\n")
print("=" * 120)
print("BEST HYPERPARAMETERS")
print("=" * 120)


print("\n1. DECISION TREE")

print(
    dt_grid.best_params_
)


print("\n2. RANDOM FOREST")

print(
    rf_grid.best_params_
)


print("\n3. BAGGING CLASSIFIER")

print(
    bag_grid.best_params_
)


# =============================================================================
# 48. FIND OVERALL BEST MODEL
# =============================================================================
#
# We select the overall best model using the 5-fold CV ROC-AUC.
#
# We do NOT select it using the test set.
#
# =============================================================================

best_model_name = (
    model_results
    .iloc[0][
        "Model"
    ]
)


if best_model_name == "Decision Tree":

    overall_best_model = (
        best_dt
    )

    overall_best_params = (
        dt_grid.best_params_
    )

    overall_best_cv_auc = (
        dt_grid.best_score_
    )

    overall_test_auc = (
        dt_auc
    )

    overall_test_accuracy = (
        dt_accuracy
    )

    overall_precision = (
        dt_precision
    )

    overall_recall = (
        dt_recall
    )

    overall_f1 = (
        dt_f1
    )

    overall_specificity = (
        dt_specificity
    )


elif best_model_name == "Random Forest":

    overall_best_model = (
        best_rf
    )

    overall_best_params = (
        rf_grid.best_params_
    )

    overall_best_cv_auc = (
        rf_grid.best_score_
    )

    overall_test_auc = (
        rf_auc
    )

    overall_test_accuracy = (
        rf_accuracy
    )

    overall_precision = (
        rf_precision
    )

    overall_recall = (
        rf_recall
    )

    overall_f1 = (
        rf_f1
    )

    overall_specificity = (
        rf_specificity
    )


else:

    overall_best_model = (
        best_bag
    )

    overall_best_params = (
        bag_grid.best_params_
    )

    overall_best_cv_auc = (
        bag_grid.best_score_
    )

    overall_test_auc = (
        bag_auc
    )

    overall_test_accuracy = (
        bag_accuracy
    )

    overall_precision = (
        bag_precision
    )

    overall_recall = (
        bag_recall
    )

    overall_f1 = (
        bag_f1
    )

    overall_specificity = (
        bag_specificity
    )


# =============================================================================
# 49. PRINT OVERALL BEST MODEL
# =============================================================================

print("\n")
print("=" * 120)
print("OVERALL BEST MODEL")
print("=" * 120)


print(
    "Best Model:",
    best_model_name
)


print(
    "\nBest Hyperparameters:"
)


print(
    overall_best_params
)


print(
    "\nBest 5-Fold CV ROC-AUC:",
    round(
        overall_best_cv_auc,
        6
    )
)


print(
    "Test Accuracy:",
    round(
        overall_test_accuracy,
        6
    )
)


print(
    "Test Precision:",
    round(
        overall_precision,
        6
    )
)


print(
    "Test Recall:",
    round(
        overall_recall,
        6
    )
)


print(
    "Test Specificity:",
    round(
        overall_specificity,
        6
    )
)


print(
    "Test F1 Score:",
    round(
        overall_f1,
        6
    )
)


print(
    "Test ROC-AUC:",
    round(
        overall_test_auc,
        6
    )
)


print("\nBest Model Object:")


print(
    overall_best_model
)


# =============================================================================
# 50. RUN EXPLICIT 5-FOLD CROSS VALIDATION ON BEST MODEL
# =============================================================================
#
# GridSearchCV already performed 5-fold CV.
#
# This additional step prints the individual fold scores of the
# final selected model.
#
# =============================================================================

print("\n")
print("=" * 120)
print("BEST MODEL - INDIVIDUAL 5-FOLD ROC-AUC SCORES")
print("=" * 120)


best_model_fold_scores = (
    cross_val_score(

        overall_best_model,

        X_train,

        y_train,

        cv=cv,

        scoring="roc_auc",

        n_jobs=-1
    )
)


# =============================================================================
# 51. PRINT EACH FOLD
# =============================================================================

for fold_number, fold_score in enumerate(

    best_model_fold_scores,

    start=1
):

    print(

        f"Fold {fold_number} ROC-AUC:",

        round(
            fold_score,
            6
        )
    )


# =============================================================================
# 52. MEAN CV SCORE
# =============================================================================

print(
    "\nMean 5-Fold ROC-AUC:",
    round(
        best_model_fold_scores.mean(),
        6
    )
)


# =============================================================================
# 53. STANDARD DEVIATION
# =============================================================================

print(
    "Standard Deviation:",
    round(
        best_model_fold_scores.std(),
        6
    )
)


# =============================================================================
# 54. MINIMUM FOLD SCORE
# =============================================================================

print(
    "Minimum Fold ROC-AUC:",
    round(
        best_model_fold_scores.min(),
        6
    )
)


# =============================================================================
# 55. MAXIMUM FOLD SCORE
# =============================================================================

print(
    "Maximum Fold ROC-AUC:",
    round(
        best_model_fold_scores.max(),
        6
    )
)


# =============================================================================
# 56. CREATE FOLD RESULTS TABLE
# =============================================================================

fold_results = pd.DataFrame({

    "Fold": [
        "Fold 1",
        "Fold 2",
        "Fold 3",
        "Fold 4",
        "Fold 5"
    ],

    "ROC-AUC": (
        best_model_fold_scores
    )
})


print("\n5-Fold Results Table:")


print(
    fold_results
    .round(6)
    .to_string(
        index=False
    )
)


# =============================================================================
# 57. FEATURE IMPORTANCE - DECISION TREE
# =============================================================================

print("\n")
print("=" * 120)
print("DECISION TREE FEATURE IMPORTANCE")
print("=" * 120)


dt_feature_importance = pd.DataFrame({

    "Feature": (
        X.columns
    ),

    "Importance": (
        best_dt.feature_importances_
    )
})


dt_feature_importance = (

    dt_feature_importance

    .sort_values(

        by="Importance",

        ascending=False
    )
)


print(
    dt_feature_importance
    .head(30)
    .to_string(
        index=False
    )
)


# =============================================================================
# 58. RANDOM FOREST FEATURE IMPORTANCE
# =============================================================================

print("\n")
print("=" * 120)
print("RANDOM FOREST FEATURE IMPORTANCE")
print("=" * 120)


rf_feature_importance = pd.DataFrame({

    "Feature": (
        X.columns
    ),

    "Importance": (
        best_rf.feature_importances_
    )
})


rf_feature_importance = (

    rf_feature_importance

    .sort_values(

        by="Importance",

        ascending=False
    )
)


print(
    rf_feature_importance
    .head(30)
    .to_string(
        index=False
    )
)


# =============================================================================
# 59. ENGINEERED FEATURE IMPORTANCE
# =============================================================================
#
# Earlier feature engineering code created new_columns.
#
# If that variable exists, we isolate the importance of only
# the engineered variables.
#
# =============================================================================

if "new_columns" in globals():

    engineered_feature_importance = (

        rf_feature_importance[

            rf_feature_importance[
                "Feature"
            ].isin(
                new_columns
            )

        ]

        .copy()
    )


    print("\n")
    print("=" * 120)
    print("RANDOM FOREST - ENGINEERED FEATURE IMPORTANCE")
    print("=" * 120)


    print(
        engineered_feature_importance
        .to_string(
            index=False
        )
    )


# =============================================================================
# 60. PRINT TOP 10 RANDOM FOREST FEATURES
# =============================================================================

print("\n")
print("=" * 120)
print("TOP 10 MOST IMPORTANT RANDOM FOREST FEATURES")
print("=" * 120)


top_10_features = (
    rf_feature_importance
    .head(10)
)


print(
    top_10_features
    .to_string(
        index=False
    )
)


# =============================================================================
# 61. OVERFITTING ANALYSIS
# =============================================================================

print("\n")
print("=" * 120)
print("OVERFITTING ANALYSIS")
print("=" * 120)


overfitting_results = (

    model_results[

        [
            "Model",
            "Train Accuracy",
            "Test Accuracy",
            "Accuracy Gap"
        ]

    ]

    .copy()
)


print(
    overfitting_results
    .round(6)
    .to_string(
        index=False
    )
)


# =============================================================================
# 62. INTERPRET OVERFITTING
# =============================================================================

print("\nOverfitting Interpretation:")


for index, row in overfitting_results.iterrows():

    gap = (
        row[
            "Accuracy Gap"
        ]
    )

    model_name = (
        row[
            "Model"
        ]
    )


    if gap < 0.02:

        print(
            model_name,
            "-> Low train/test accuracy gap."
        )


    elif gap < 0.05:

        print(
            model_name,
            "-> Moderate train/test accuracy gap."
        )


    else:

        print(
            model_name,
            "-> Large train/test accuracy gap; check for overfitting."
        )


# =============================================================================
# 63. COMPARE CV ROC-AUC VS TEST ROC-AUC
# =============================================================================

model_results[
    "CV-Test AUC Gap"
] = (

    model_results[
        "Best CV ROC-AUC"
    ]

    -

    model_results[
        "Test ROC-AUC"
    ]
).abs()


print("\n")
print("=" * 120)
print("CV ROC-AUC VS TEST ROC-AUC")
print("=" * 120)


print(
    model_results[

        [
            "Model",
            "Best CV ROC-AUC",
            "Test ROC-AUC",
            "CV-Test AUC Gap"
        ]

    ]
    .round(6)
    .to_string(
        index=False
    )
)


# =============================================================================
# 64. SAVE BEST MODEL
# =============================================================================

best_model_path = (

    "/content/drive/MyDrive/Home Credit/"
    "best_home_credit_model.pkl"
)


joblib.dump(

    overall_best_model,

    best_modelBaggingClassifier(

    estimator=base_tree,

    random_state=42,

    n_jobs=-1
)


# =============================================================================
# 35. BAGGING PARAMETER GRID
# =============================================================================

bag_param_grid = {

    "n_estimators": [
        25,
        50
    ],

    "max_samples": [
        0.7,
        1.0
    ],

    "max_features": [
        0.7,
        1.0
    ],

    "bootstrap": [
        True
    ],

    "estimator__max_depth": [
        5,
        10
    ],

    "estimator__min_samples_split": [
        10,
        20
    ],

    "estimator__min_samples_leaf": [
        5,
        10
    ],

    "estimator__class_weight": [
        None,
        "balanced"
    ]
}


print("\n" + "=" * 120)
print("BAGGING CLASSIFIER GRID SEARCH")
print("=" * 120)

print("Parameter Grid:")

print(
    bag_param_grid
)


# =============================================================================
# 36. BAGGING GRID SEARCH CV
# =============================================================================

bag_grid = GridSearchCV(

    estimator=bagging,

    param_grid=bag_param_grid,

    scoring="roc_auc",

    cv=cv,

    n_jobs=-1,

    verbose=1,

    return_train_score=True,

    refit=True
)


bag_start_time = (
    time.time()
)


bag_grid.fit(

    X_train,

    y_train
)


bag_grid_time = (

    time.time()

    -

    bag_start_time
)


# =============================================================================
# 37. BEST BAGGING MODEL
# =============================================================================

best_bag = (
    bag_grid.best_estimator_
)


print("\nBest Bagging Parameters:")

print(
    bag_grid.best_params_
)


print(
    "\nBest Bagging CV ROC-AUC:",
    round(
        bag_grid.best_score_,
        6
    )
)


print(
    "Grid Search Time:",
    round(
        bag_grid_time,
        2
    ),
    "seconds"
)


# =============================================================================
# 38. BAGGING PREDICTIONS
# =============================================================================

bag_train_prediction = (
    best_bag.predict(
        X_train
    )
)


bag_test_prediction = (
    best_bag.predict(
        X_test
    )
)


bag_test_probability = (
    best_bag.predict_proba(
        X_test
    )[:, 1]
)


# =============================================================================
# 39. BAGGING METRICS
# =============================================================================

bag_train_accuracy = (
    accuracy_score(
        y_train,
        bag_train_prediction
    )
)


bag_accuracy = (
    accuracy_score(
        y_test,
        bag_test_prediction
    )
)


bag_balanced_accuracy = (
    balanced_accuracy_score(
        y_test,
        bag_test_prediction
    )
)


bag_precision = (
    precision_score(
        y_test,
        bag_test_prediction,
        zero_division=0
    )
)


bag_recall = (
    recall_score(
        y_test,
        bag_test_prediction,
        zero_division=0
    )
)


bag_f1 = (
    f1_score(
        y_test,
        bag_test_prediction,
        zero_division=0
    )
)


bag_auc = (
    roc_auc_score(
        y_test,
        bag_test_probability
    )
)


# =============================================================================
# 40. BAGGING CONFUSION MATRIX
# =============================================================================

bag_confusion_matrix = (
    confusion_matrix(
        y_test,
        bag_test_prediction
    )
)


bag_tn, bag_fp, bag_fn, bag_tp = (
    bag_confusion_matrix.ravel()
)


bag_specificity = (

    bag_tn

    /

    (
        bag_tn
        +
        bag_fp
    )
)


# =============================================================================
# 41. PRINT BAGGING RESULTS
# =============================================================================

print("\n" + "=" * 120)
print("BEST BAGGING CLASSIFIER - FINAL RESULTS")
print("=" * 120)


print(
    "Training Accuracy :",
    round(
        bag_train_accuracy,
        6
    )
)


print(
    "Testing Accuracy  :",
    round(
        bag_accuracy,
        6
    )
)


print(
    "Balanced Accuracy :",
    round(
        bag_balanced_accuracy,
        6
    )
)


print(
    "Precision         :",
    round(
        bag_precision,
        6
    )
)


print(
    "Recall            :",
    round(
        bag_recall,
        6
    )
)


print(
    "Specificity       :",
    round(
        bag_specificity,
        6
    )
)


print(
    "F1 Score          :",
    round(
        bag_f1,
        6
    )
)


print(
    "ROC-AUC           :",
    round(
        bag_auc,
        6
    )
)


print("\nConfusion Matrix:")


print(
    bag_confusion_matrix
)


print(
    "\nTrue Negative:",
    bag_tn
)


print(
    "False Positive:",
    bag_fp
)


print(
    "False Negative:",
    bag_fn
)


print(
    "True Positive:",
    bag_tp
)


print("\nClassification Report:")


print(
    classification_report(

        y_test,

        bag_test_prediction,

        digits=4,

        zero_division=0
    )
)


# =============================================================================
# 42. BAGGING GRID SEARCH RESULTS
# =============================================================================

bag_cv_results = (
    pd.DataFrame(
        bag_grid.cv_results_
    )
)


bag_cv_results = (

    bag_cv_results[

        [
            "params",
            "mean_train_score",
            "mean_test_score",
            "std_test_score",
            "rank_test_score"
        ]

    ]

    .sort_values(
        by="rank_test_score"
    )
)


print("\nTop 10 Bagging Parameter Combinations:")


print(
    bag_cv_results
    .head(10)
    .to_string(
        index=False
    )
)


# =============================================================================
# 43. CREATE FINAL MODEL COMPARISON
# =============================================================================

model_results = pd.DataFrame({

    "Model": [

        "Decision Tree",

        "Random Forest",

        "Bagging Classifier"
    ],


    "Best CV ROC-AUC": [

        dt_grid.best_score_,

        rf_grid.best_score_,

        bag_grid.best_score_
    ],


    "Train Accuracy": [

        dt_train_accuracy,

        rf_train_accuracy,

        bag_train_accuracy
    ],


    "Test Accuracy": [

        dt_accuracy,

        rf_accuracy,

        bag_accuracy
    ],


    "Balanced Accuracy": [

        dt_balanced_accuracy,

        rf_balanced_accuracy,

        bag_balanced_accuracy
    ],


    "Precision": [

        dt_precision,

        rf_precision,

        bag_precision
    ],


    "Recall": [

        dt_recall,

        rf_recall,

        bag_recall
    ],


    "Specificity": [

        dt_specificity,

        rf_specificity,

        bag_specificity
    ],


    "F1 Score": [

        dt_f1,

        rf_f1,

        bag_f1
    ],


    "Test ROC-AUC": [

        dt_auc,

        rf_auc,

        bag_auc
    ],


    "Grid Search Time": [

        dt_grid_time,

        rf_grid_time,

        bag_grid_time
    ]
})


# =============================================================================
# 44. ADD OVERFITTING GAP
# =============================================================================

model_results[
    "Accuracy Gap"
] = (

    model_results[
        "Train Accuracy"
    ]

    -

    model_results[
        "Test Accuracy"
    ]
)


# =============================================================================
# 45. SORT MODELS ACCORDING TO CV ROC-AUC
# =============================================================================

model_results = (
    model_results
    .sort_values(

        by="Best CV ROC-AUC",

        ascending=False
    )
    .reset_index(
        drop=True
    )
)


# =============================================================================
# 46. PRINT FINAL COMPARISON
# =============================================================================

print("\n")
print("=" * 140)
print("FINAL MODEL COMPARISON")
print("=" * 140)


print(
    model_results
    .round(6)
    .to_string(
        index=False
    )
)


# =============================================================================
# 47. PRINT BEST PARAMETERS OF ALL MODELS
# =============================================================================

print("\n")
print("=" * 120)
print("BEST HYPERPARAMETERS")
print("=" * 120)


print("\n1. DECISION TREE")

print(
    dt_grid.best_params_
)


print("\n2. RANDOM FOREST")

print(
    rf_grid.best_params_
)


print("\n3. BAGGING CLASSIFIER")

print(
    bag_grid.best_params_
)


# =============================================================================
# 48. FIND OVERALL BEST MODEL
# =============================================================================
#
# We select the overall best model using the 5-fold CV ROC-AUC.
#
# We do NOT select it using the test set.
#
# =============================================================================

best_model_name = (
    model_results
    .iloc[0][
        "Model"
    ]
)


if best_model_name == "Decision Tree":

    overall_best_model = (
        best_dt
    )

    overall_best_params = (
        dt_grid.best_params_
    )

    overall_best_cv_auc = (
        dt_grid.best_score_
    )

    overall_test_auc = (
        dt_auc
    )

    overall_test_accuracy = (
        dt_accuracy
    )

    overall_precision = (
        dt_precision
    )

    overall_recall = (
        dt_recall
    )

    overall_f1 = (
        dt_f1
    )

    overall_specificity = (
        dt_specificity
    )


elif best_model_name == "Random Forest":

    overall_best_model = (
        best_rf
    )

    overall_best_params = (
        rf_grid.best_params_
    )

    overall_best_cv_auc = (
        rf_grid.best_score_
    )

    overall_test_auc = (
        rf_auc
    )

    overall_test_accuracy = (
        rf_accuracy
    )

    overall_precision = (
        rf_precision
    )

    overall_recall = (
        rf_recall
    )

    overall_f1 = (
        rf_f1
    )

    overall_specificity = (
        rf_specificity
    )


else:

    overall_best_model = (
        best_bag
    )

    overall_best_params = (
        bag_grid.best_params_
    )

    overall_best_cv_auc = (
        bag_grid.best_score_
    )

    overall_test_auc = (
        bag_auc
    )

    overall_test_accuracy = (
        bag_accuracy
    )

    overall_precision = (
        bag_precision
    )

    overall_recall = (
        bag_recall
    )

    overall_f1 = (
        bag_f1
    )

    overall_specificity = (
        bag_specificity
    )


# =============================================================================
# 49. PRINT OVERALL BEST MODEL
# =============================================================================

print("\n")
print("=" * 120)
print("OVERALL BEST MODEL")
print("=" * 120)


print(
    "Best Model:",
    best_model_name
)


print(
    "\nBest Hyperparameters:"
)


print(
    overall_best_params
)


print(
    "\nBest 5-Fold CV ROC-AUC:",
    round(
        overall_best_cv_auc,
        6
    )
)


print(
    "Test Accuracy:",
    round(
        overall_test_accuracy,
        6
    )
)


print(
    "Test Precision:",
    round(
        overall_precision,
        6
    )
)


print(
    "Test Recall:",
    round(
        overall_recall,
        6
    )
)


print(
    "Test Specificity:",
    round(
        overall_specificity,
        6
    )
)


print(
    "Test F1 Score:",
    round(
        overall_f1,
        6
    )
)


print(
    "Test ROC-AUC:",
    round(
        overall_test_auc,
        6
    )
)


print("\nBest Model Object:")


print(
    overall_best_model
)


# =============================================================================
# 50. RUN EXPLICIT 5-FOLD CROSS VALIDATION ON BEST MODEL
# =============================================================================
#
# GridSearchCV already performed 5-fold CV.
#
# This additional step prints the individual fold scores of the
# final selected model.
#
# =============================================================================

print("\n")
print("=" * 120)
print("BEST MODEL - INDIVIDUAL 5-FOLD ROC-AUC SCORES")
print("=" * 120)


best_model_fold_scores = (
    cross_val_score(

        overall_best_model,

        X_train,

        y_train,

        cv=cv,

        scoring="roc_auc",

        n_jobs=-1
    )
)


# =============================================================================
# 51. PRINT EACH FOLD
# =============================================================================

for fold_number, fold_score in enumerate(

    best_model_fold_scores,

    start=1
):

    print(

        f"Fold {fold_number} ROC-AUC:",

        round(
            fold_score,
            6
        )
    )


# =============================================================================
# 52. MEAN CV SCORE
# =============================================================================

print(
    "\nMean 5-Fold ROC-AUC:",
    round(
        best_model_fold_scores.mean(),
        6
    )
)


# =============================================================================
# 53. STANDARD DEVIATION
# =============================================================================

print(
    "Standard Deviation:",
    round(
        best_model_fold_scores.std(),
        6
    )
)


# =============================================================================
# 54. MINIMUM FOLD SCORE
# =============================================================================

print(
    "Minimum Fold ROC-AUC:",
    round(
        best_model_fold_scores.min(),
        6
    )
)


# =============================================================================
# 55. MAXIMUM FOLD SCORE
# =============================================================================

print(
    "Maximum Fold ROC-AUC:",
    round(
        best_model_fold_scores.max(),
        6
    )
)


# =============================================================================
# 56. CREATE FOLD RESULTS TABLE
# =============================================================================

fold_results = pd.DataFrame({

    "Fold": [
        "Fold 1",
        "Fold 2",
        "Fold 3",
        "Fold 4",
        "Fold 5"
    ],

    "ROC-AUC": (
        best_model_fold_scores
    )
})


print("\n5-Fold Results Table:")


print(
    fold_results
    .round(6)
    .to_string(
        index=False
    )
)


# =============================================================================
# 57. FEATURE IMPORTANCE - DECISION TREE
# =============================================================================

print("\n")
print("=" * 120)
print("DECISION TREE FEATURE IMPORTANCE")
print("=" * 120)


dt_feature_importance = pd.DataFrame({

    "Feature": (
        X.columns
    ),

    "Importance": (
        best_dt.feature_importances_
    )
})


dt_feature_importance = (

    dt_feature_importance

    .sort_values(

        by="Importance",

        ascending=False
    )
)


print(
    dt_feature_importance
    .head(30)
    .to_string(
        index=False
    )
)


# =============================================================================
# 58. RANDOM FOREST FEATURE IMPORTANCE
# =============================================================================

print("\n")
print("=" * 120)
print("RANDOM FOREST FEATURE IMPORTANCE")
print("=" * 120)


rf_feature_importance = pd.DataFrame({

    "Feature": (
        X.columns
    ),

    "Importance": (
        best_rf.feature_importances_
    )
})


rf_feature_importance = (

    rf_feature_importance

    .sort_values(

        by="Importance",

        ascending=False
    )
)


print(
    rf_feature_importance
    .head(30)
    .to_string(
        index=False
    )
)


# =============================================================================
# 59. ENGINEERED FEATURE IMPORTANCE
# =============================================================================
#
# Earlier feature engineering code created new_columns.
#
# If that variable exists, we isolate the importance of only
# the engineered variables.
#
# =============================================================================

if "new_columns" in globals():

    engineered_feature_importance = (

        rf_feature_importance[

            rf_feature_importance[
                "Feature"
            ].isin(
                new_columns
            )

        ]

        .copy()
    )


    print("\n")
    print("=" * 120)
    print("RANDOM FOREST - ENGINEERED FEATURE IMPORTANCE")
    print("=" * 120)


    print(
        engineered_feature_importance
        .to_string(
            index=False
        )
    )


# =============================================================================
# 60. PRINT TOP 10 RANDOM FOREST FEATURES
# =============================================================================

print("\n")
print("=" * 120)
print("TOP 10 MOST IMPORTANT RANDOM FOREST FEATURES")
print("=" * 120)


top_10_features = (
    rf_feature_importance
    .head(10)
)


print(
    top_10_features
    .to_string(
        index=False
    )
)


# =============================================================================
# 61. OVERFITTING ANALYSIS
# =============================================================================

print("\n")
print("=" * 120)
print("OVERFITTING ANALYSIS")
print("=" * 120)


overfitting_results = (

    model_results[

        [
            "Model",
            "Train Accuracy",
            "Test Accuracy",
            "Accuracy Gap"
        ]

    ]

    .copy()
)


print(
    overfitting_results
    .round(6)
    .to_string(
        index=False
    )
)


# =============================================================================
# 62. INTERPRET OVERFITTING
# =============================================================================

print("\nOverfitting Interpretation:")


for index, row in overfitting_results.iterrows():

    gap = (
        row[
            "Accuracy Gap"
        ]
    )

    model_name = (
        row[
            "Model"
        ]
    )


    if gap < 0.02:

        print(
            model_name,
            "-> Low train/test accuracy gap."
        )


    elif gap < 0.05:

        print(
            model_name,
            "-> Moderate train/test accuracy gap."
        )


    else:

        print(
            model_name,
            "-> Large train/test accuracy gap; check for overfitting."
        )


# =============================================================================
# 63. COMPARE CV ROC-AUC VS TEST ROC-AUC
# =============================================================================

model_results[
    "CV-Test AUC Gap"
] = (

    model_results[
        "Best CV ROC-AUC"
    ]

    -

    model_results[
        "Test ROC-AUC"
    ]
).abs()


print("\n")
print("=" * 120)
print("CV ROC-AUC VS TEST ROC-AUC")
print("=" * 120)


print(
    model_results[

        [
            "Model",
            "Best CV ROC-AUC",
            "Test ROC-AUC",
            "CV-Test AUC Gap"
        ]

    ]
    .round(6)
    .to_string(
        index=False
    )
)


# =============================================================================
# 64. SAVE BEST MODEL
# =============================================================================

best_model_path = (

    "/content/drive/MyDrive/Home Credit/"
    "best_home_credit_model.pkl"
)


joblib.dump(

    overall_best_model,

    best_model
```
