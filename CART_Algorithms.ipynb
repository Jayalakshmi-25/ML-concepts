```python
# ============================================================
# DECISION TREE vs RANDOM FOREST vs BAGGING CLASSIFICATION
# Dataset: Wine Dataset
# ============================================================

# 1. Import Libraries
import pandas as pd

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, BaggingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)


# ============================================================
# 2. Load Dataset
# ============================================================

wine = load_wine()

X = pd.DataFrame(
    wine.data,
    columns=wine.feature_names
)

y = pd.Series(
    wine.target,
    name="Wine_Class"
)

print("Dataset Shape:", X.shape)

print("\nFirst 5 Rows:")
print(X.head())

print("\nTarget Classes:")
print(wine.target_names)

print("\nClass Distribution:")
print(y.value_counts())


# ============================================================
# 3. Train-Test Split
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Data:", X_train.shape)
print("Testing Data :", X_test.shape)


# ============================================================
# 4. DECISION TREE CLASSIFIER
# ============================================================

decision_tree = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

decision_tree.fit(X_train, y_train)

dt_pred = decision_tree.predict(X_test)


# ============================================================
# 5. RANDOM FOREST CLASSIFIER
# ============================================================

random_forest = RandomForestClassifier(
    n_estimators=100,
    max_depth=5,
    random_state=42
)

random_forest.fit(X_train, y_train)

rf_pred = random_forest.predict(X_test)


# ============================================================
# 6. BAGGING CLASSIFIER
# ============================================================

# Base model used inside Bagging
base_tree = DecisionTreeClassifier(
    random_state=42
)

bagging = BaggingClassifier(
    estimator=base_tree,
    n_estimators=100,
    max_samples=0.8,
    bootstrap=True,
    random_state=42
)

bagging.fit(X_train, y_train)

bagging_pred = bagging.predict(X_test)


# ============================================================
# 7. Evaluation Function
# ============================================================

def evaluate_model(model_name, y_true, y_pred):

    accuracy = accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        average="weighted"
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="weighted"
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average="weighted"
    )

    print("\n" + "=" * 60)
    print(model_name)
    print("=" * 60)

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_true, y_pred))

    print("\nClassification Report:")
    print(classification_report(y_true, y_pred))

    return [
        model_name,
        accuracy,
        precision,
        recall,
        f1
    ]


# ============================================================
# 8. Evaluate Decision Tree
# ============================================================

dt_result = evaluate_model(
    "Decision Tree",
    y_test,
    dt_pred
)


# ============================================================
# 9. Evaluate Random Forest
# ============================================================

rf_result = evaluate_model(
    "Random Forest",
    y_test,
    rf_pred
)


# ============================================================
# 10. Evaluate Bagging
# ============================================================

bagging_result = evaluate_model(
    "Bagging Classifier",
    y_test,
    bagging_pred
)


# ============================================================
# 11. Compare All Models
# ============================================================

results = pd.DataFrame(
    [
        dt_result,
        rf_result,
        bagging_result
    ],
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ]
)

print("\n" + "=" * 60)
print("FINAL MODEL COMPARISON")
print("=" * 60)

print(results.round(4))


# ============================================================
# 12. Find Best Model
# ============================================================

best_model = results.loc[
    results["Accuracy"].idxmax()
]

print("\nBest Model:", best_model["Model"])
print("Best Accuracy:", round(best_model["Accuracy"], 4))


# ============================================================
# CONCEPTUAL DIFFERENCE
# ============================================================

# 1. DECISION TREE
#
# Training Data
#      |
#      v
# Decision Tree
#      |
#      v
# Prediction
#
# Only ONE tree is created.


# ============================================================
# 2. BAGGING
# ============================================================

# Original Training Data
#          |
#          | Bootstrap Sampling
#          |
#     -----------------
#     |       |       |
#     v       v       v
# Sample 1 Sample 2 Sample 3
#     |       |       |
#     v       v       v
#   Tree 1  Tree 2  Tree 3
#     |       |       |
#     -----------------
#             |
#             v
#      Majority Voting
#             |
#             v
#      Final Prediction


# ============================================================
# 3. RANDOM FOREST
# ============================================================

# Random Forest is also a BAGGING-based algorithm.
#
# However, Random Forest introduces additional randomness.
#
# Bagging:
# Random Rows
# +
# All Features can be considered
#
# Random Forest:
# Random Rows
# +
# Random subset of Features
#
#
#               Training Data
#                     |
#        ---------------------------
#        |            |            |
#        v            v            v
#   Bootstrap 1  Bootstrap 2  Bootstrap 3
#        |            |            |
#   Random Features Random Features Random Features
#        |            |            |
#      Tree 1       Tree 2       Tree 3
#        |            |            |
#        ---------------------------
#                     |
#                     v
#              Majority Voting
#                     |
#                     v
#              Final Prediction


# ============================================================
# SIMPLE DIFFERENCE
# ============================================================

# Decision Tree:
# One Tree

# Bagging:
# Multiple Trees
# + Bootstrap Sampling
# + Majority Voting

# Random Forest:
# Multiple Trees
# + Bootstrap Sampling
# + Random Feature Selection
# + Majority Voting
```
